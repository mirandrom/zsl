```python
from huggingface_hub import hf_hub_download
from zsl_config import ZSL_DIR_OUT_OLMO
```

```python
# convert hf_hub model size to run names used throughout codebase
MODEL_SIZE_TO_RUN = {
        '14M': '1028-rmsnorm-14m',
        '37M': '1028-rmsnorm-37m',
        '78M': '1028-rmsnorm-78m',
        '144M': '1028-rmsnorm-144m',
        '285M': '1028-rmsnorm-285m',
        '472M': '1028-rmsnorm-472m',
    }

# iterables for all model sizes and steps
model_sizes = ['14M','37M','78M','144M','285M','472M']
steps = [int(2**i) for i in range(19)]
filenames = ['config.yaml', 'model.pt', 'optim.pt', 'train.pt']

# construct run dir path
def get_run_dir(model_size, step):
    run = MODEL_SIZE_TO_RUN[model_size]
    return ZSL_DIR_OUT_OLMO / run / f"step{step}-unsharded"
```

```python
for model_size in model_sizes:
    for step in steps:
        run_dir = get_run_dir(model_size, step)
        revision = f"OLMo-{model_size}-step{step}"
        
        print(revision)
        print('-'*40)
        
        for fn in filenames:
            f = run_dir / fn
            # hf_hub_download handles already existing downloads
            hf_hub_download(
                repo_id="mirandrom/zsl-checkpoints", 
                filename=fn, 
                revision=revision,
                local_dir=run_dir,
            )
```
